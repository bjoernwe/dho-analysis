## SFA loadings

Visualizes `data/sentence_sfa_loadings.csv` — for each zero-shot label, its *effective* loading on
the first 5 slow features (see `experiments/CalcSentenceSfa.kt`). The loadings compose the whole
chain PCA -> SFA, so they express each slow feature directly in raw label space. One horizontal bar
chart per feature, bars sorted by loading value, colored on a diverging scale centered at zero.

For each slow feature, also visualizes the most/least representative sentences from
`data/sentence_sfa_features.csv` — the sentences whose score on that feature is highest and lowest.

Slow features are ordered by slowness: `sf1` is the label-space direction that changes most slowly
from practice-log post to post, `sf5` the fastest of the five.


```kotlin
%useLatestDescriptors
%use dataframe
%use kandy
```

```kotlin
val loadings = DataFrame.readCsv("../data/sentence_sfa_loadings.csv")
println("Rows: ${loadings.rowsCount()}, columns: ${loadings.columnNames()}")
loadings.head()
```

```kotlin
fun plotLoadings(component: String) = run {
    val sorted = loadings.sortBy(component)
    val labels = sorted["label"].toList().map { it.toString() }
    val values = sorted[component].toList().map { (it as Number).toDouble() }

    plot {
        layout {
            title = "Label weights in SFA component \"$component\""
            size = 700 to 1600
        }
        barsH {
            y(labels) { axis.name = "label" }
            x(values) { axis.name = "loading" }
            fillColor(values) {
                scale = continuousColorGradient2(Color.BLUE, Color.WHITE, Color.RED, 0.0)
                legend.name = "loading"
            }
        }
    }
}
```

```kotlin
val features = DataFrame.readCsv("../data/sentence_sfa_features.csv").distinctBy("sentence")
println("Rows: ${features.rowsCount()}, columns: ${features.columnNames()}")
features.head()
```

```kotlin
fun truncate(s: String, maxLength: Int = 90) =
    if (s.length <= maxLength) s else s.take(maxLength - 1) + "…"

fun printSentences(component: String, n: Int = 5) = run {
    val sorted = features.sortByDesc(component)
    val top = sorted.take(n)
    val bottom = sorted.takeLast(n)

    fun printRows(rows: AnyFrame) {
        val sentences = rows["sentence"].toList().map { truncate(it.toString()) }
        val values = rows[component].toList().map { (it as Number).toDouble() }
        sentences.zip(values).forEach { (sentence, value) ->
            println("%+.3f  %s".format(value, sentence))
        }
    }

    printRows(top)
    println("...")
    printRows(bottom)
}
```
